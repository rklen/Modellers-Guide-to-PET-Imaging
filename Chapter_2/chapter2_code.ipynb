```python
import numpy as np
#vectors, 2D matrices, and larger arrays can be created with np.array
vector=np.array([1,3,5])
print('an example vector of 3 elements:',vector)
matrix=np.array([[3,1],[4,-7],[6,3]])
print('an example 3*2 matrix:')
print(matrix)
array3d=np.array([[[3,1],[4,-7],[6,3]],[[5,6],[8,0],[9,1]],[[-3,7],[-4,6],
    [-6,5]]])
print('an example 3*3*2 array:')
print(array3d)
#to check the dimensions of an array, we use .shape
print('dimensions of the example matrix:',matrix.shape)
print('dimensions of the example 3D array:',array3d.shape)
#to locate the element at the location, we use indices
#recall that the indexing in Python starts at zero!
print('The element on the 2nd row and the 1st column of the example matrix:',matrix[1,0])
print('The element at the location (3,2,1) of the example array:',array3d[2,1,0])
#so matrix[1,0] is the element on the second row and the first column
#we can change their values like this
matrix[1,0]=0
print(matrix)
#we can initialise arrays full of zeros or ones easily
#we just have to decide the desired shape
zeroMatrix=np.zeros((3,4))
print(zeroMatrix)
arrayOfOnes=np.ones((2,2,4))
print(arrayOfOnes)
#we can create a binary matrix based on some condition, for instance
binaryMatrix=np.array(matrix>0,dtype=int)
print(binaryMatrix)
#initialise matrices A and B
A=np.array([[3,1],[4,-7],[6,3]])
B=np.array([[7,8],[-1,6],[0,4]])
print('A:')
print(A)
print('B:')
print(B)
#some basic operations:
print('The sum of all elements in A:',np.sum(A))
print('The smallest element in A:',np.min(A))
print('The greatest element in A:',np.max(A))
print('The element-wise minimum of A and B:')
print(np.minimum(A,B))
print('The element-wise maximum of A and B:')
print(np.maximum(A,B))
print('The sum of A and B:')
print(A+B)
print('Adding the number 1 to all elements of A:')
print(A+1)
print('The difference A minus B')
print(A-B)
print('The absolute values of this difference between A and B')
print(abs(A-B))
print('The element-wise product between A and B:')
print(A*B)
print('The product 3A:')
print(3*A)
print('The dot product of [1,3,-5] and [-9,0,4]:')
print(np.dot(np.array([1,3,-5]),np.array([-9,0,4])))
#initialise a third matrix called C
C=np.array([[3,1,0],[4,-7,3],[6,3,0],[7,1,1]])
print('C:')
print(C)
print('Matrix product of C and B')
print(np.matmul(C,B))
print('Transpose of A:')
print(np.transpose(A))
```

```python
import numpy as np
import matplotlib.pyplot as plt
#choose some points for the x- and the y-coordinates (here, x_points=0,0.1,0.2,...,3.0, and y_i=x_i^3)
x_points=np.arange(0,3.1,0.1)
y_points=x_points**3
#initialise an empty array for the approximated derivative
deriv_approx=np.full(shape=len(x_points),fill_value=None)
#fill the points of deriv_approx in a loop
for i in range(len(x_points)-1):
    deriv_approx[i]=(y_points[i+1]-y_points[i])/(x_points[i+1]-x_points[i])
#count the exact derivative (i.e. 3*x^2 for x^3)
deriv_exact=3*x_points**2
#plot the function and its derivative from the points
plt.plot(x_points,y_points,color='black')
plt.plot(x_points,deriv_approx,'b--')
plt.plot(x_points,deriv_exact,color='lightblue')
plt.legend(['Original function','Approx. derivative','Exact derivative'])
```
